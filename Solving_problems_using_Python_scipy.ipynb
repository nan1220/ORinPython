## Example 1: Factory Production Problem

Maximize revenue: $40x_1 + 50x_2$

Subject to:
- Wood constraint: $x_1 + 2x_2 \leq 40$
- Labor constraint: $4x_1 + 3x_2 \leq 120$
- Non-negativity: $x_1, x_2 \geq 0$

where $x_1$ = chairs, $x_2$ = tables

```python
# Define the problem
# scipy.optimize.linprog minimizes, so we negate coefficients for maximization
c = [-40, -50]  # Objective coefficients (negated for maximization)

# Inequality constraints: A_ub @ x <= b_ub
A_ub = [[1, 2],   # wood constraint
        [4, 3]]   # labor constraint
b_ub = [40, 120]

# Variable bounds (x >= 0)
bounds = [(0, None), (0, None)]

# Solve the linear program
result = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

# Display results
print(f"chairs: {result.x[0]:.1f}")
print(f"tables: {result.x[1]:.1f}")
print(f"optimal total revenue: {-result.fun:.1f}")  # Negate back for max problem
```

Given the constraints, the maximum revenue is $1360 structured such that we produce 24 chairs and 8 tables.
